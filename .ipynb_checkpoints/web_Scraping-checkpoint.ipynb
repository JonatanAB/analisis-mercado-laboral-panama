```python
import re

def es_reciente(texto_fecha):
    texto = texto_fecha.lower()
    if "horas" in texto or "hora" in texto:
        return True
    if "ayer" in texto:
        return True
    match = re.search(r"hace (\d+) d", texto)
    if match:
        dias = int(match.group(1))
        if dias <= 15:
            return True
    return False

print(es_reciente("Publicado hace 3 horas"))
print(es_reciente("Publicado ayer"))
print(es_reciente("Publicado hace 10 días"))
print(es_reciente("Publicado hace 20 días"))
print(es_reciente("Actualizado hace 5 días"))
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

def crear_driver():
    options = webdriver.ChromeOptions()
    options.binary_location = r"C:\Program Files\BraveSoftware\Brave-Browser\Application\brave.exe"
    options.add_argument("--no-sandbox")
    options.add_argument("--disable-dev-shm-usage")
    options.add_argument("--disable-blink-features=AutomationControlled")
    options.add_experimental_option("excludeSwitches", ["enable-automation"])
    return webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)
```

```python
datos = []
driver = crear_driver()

for pagina in range(1, 43):
    try:
        if pagina % 10 == 0:
            print(f"Reiniciando driver en página {pagina}...")
            driver.quit()
            time.sleep(3)
            driver = crear_driver()
        print(f"Scrapeando página {pagina} de 42...")
        
        url = f"https://www.konzerta.com/empleos-publicacion-menor-a-15-dias.html?recientes=true&page={pagina}"
        driver.set_page_load_timeout(30)
        driver.get(url)
        time.sleep(7)
        
        html = driver.page_source
        soup = BeautifulSoup(html, "html.parser")
        
        tarjetas = soup.find_all("div", class_=lambda x: x and "sc-iNwVbF" in x)
    
        for tarjeta in tarjetas:
            h2 = tarjeta.find("h2")
            h3s = tarjeta.find_all("h3")
            
            titulo = h2.text.strip() if h2 else "No encontrado"
            fecha = h3s[0].text.strip() if h3s else "No encontrado"
            empresa = h3s[1].text.strip() if len(h3s) > 1 else "No encontrado"
            modalidad = h3s[-1].text.strip() if h3s else "No encontrado"
            
            try:
                float(h3s[2].text.strip())
                ubicacion = h3s[3].text.strip() if len(h3s) > 3 else "No encontrado"
            except:
                ubicacion = h3s[2].text.strip() if len(h3s) > 2 else "No encontrado"
            
            datos.append({
                "titulo": titulo,
                "empresa": empresa,
                "ubicacion": ubicacion,
                "modalidad": modalidad,
                "fecha": fecha
            })
            
            print(f"  Acumulado: {len(datos)} ofertas")
    except Exception as e:
        print(f"Error en página {pagina}: {e}, continuando...")
        time.sleep(5)
        continue
driver.quit()
```

```python
df = pd.DataFrame(datos)
print(f"\nCantidad de ofertas: {len(datos)}")
df.head(10)
```

```python
df.to_csv("Konzerta_ofertas_laborales.csv", index=False, encoding = "utf-8-sig")
print("Archivo guardado exitosamente")
```

```python
def categorizar(titulo):
    titulo=titulo.lower()

    if any(x in titulo for x in ["call center", "bilingual", "customer service", "agente", "Atención al Cliente"]):
        return "Call Center / Atención al Cliente"
    elif any(x in titulo for x in ["sistemas", "software", "developer", "programador", "desarrollo", "frontend", "backend", "fullstack","apps", "app","aplicacion", "aplicaciones", "web", "IT", "tecnología", "TI"]):
        return "Tecnologia / Desarrollo"
    elif any(x in titulo for x in ["analista de datos", "data analyst", "data engineer", "ingeniero de datos", "business intelligence", "big data", "analytics", "gobierno de datos", "arquitecto de datos"]):
        return "Area de Datos"
    elif any(x in titulo for x in ["contab", "contador", "contable", "oficinista de contab"]):
        return "Contabilidad"
    elif any(x in titulo for x in ["asistente", "assistant", "auxiliar"]):
        return "Asistente / Auxiliar"
    elif any(x in titulo for x in ["ingenier", "engineer"]):
        return "Ingeniería"
    elif any(x in titulo for x in ["conductor", "chofer", "driver"]):
        return "Conductores"
    elif any(x in titulo for x in ["supervisor", "coordinador", "jefe", "gerente", "manager"]):
        return "Supervisión / Gerencia"
    elif any(x in titulo for x in ["enfermera", "médico", "salud", "farmac","medico","doctor"]):
        return "Salud"
    elif any(x in titulo for x in ["cajero", "cajera"]):
        return "Caja / Finanzas"
    elif any(x in titulo for x in ["mecánico", "mecanico", "automotriz"]):
        return "Mecánica / Automotriz"
    elif any(x in titulo for x in ["cocinero", "bartender", "salonero", "cocina"]):
        return "Gastronomía / Hostelería"
    elif any(x in titulo for x in ["ayudante", "almacenista", "estibador", "operador", "mensajero"]):
        return "Operativo / Logística"
    elif any(x in titulo for x in ["calidad", "riesgo", "control"]):
        return "Calidad / Riesgo"
    elif any(x in titulo for x in ["docente", "profesor", "educador", "instructor", "entrenador"]):
        return "Educación / Capacitación"
    elif any(x in titulo for x in ["abogado", "legal", "jurídico"]):
        return "Legal"
    elif any(x in titulo for x in ["asesor", "seguros", "crédito", "credito"]):
        return "Banca / Seguros"
    else:
        return "Otros"
        
df["Categoria"] = df["titulo"].apply(categorizar)
print(df["Categoria"].value_counts())  
```
